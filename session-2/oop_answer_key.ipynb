# Part A: Guide

## 1. Class and object
- A **class** is a **blueprint** that describes what something *has* (its data) and what it *can do* (its actions).
- An **object** is one real thing made from that blueprint.

> 🍪 **Analogy:** the class is a **cookie cutter** and objects are **cookies**. One cutter can make many cookies, and each cookie can have different toppings.

```python
class Cookie:
    pass                 # an empty blueprint (for now)

c1 = Cookie()            # make an object
c2 = Cookie()            # make another one
print(c1)
print(c2)                # two different objects
```

## 2. `__init__` and `self`
- `__init__` runs **automatically when you create an object**. Use it to set the object's starting data, which are called its **attributes**.
- `self` means **"this object"**. It's how each object stores and finds *its own* data.

> 🏷️ **Analogy:** `__init__` is filling in a **new student's registration form**. `self` is the name on that form, so the data goes to the right student.

```python
class Student:
    def __init__(self, name, age):
        self.name = name     # save the data ON this object
        self.age = age

ana = Student("Ana", 22)
bo = Student("Bo", 25)

print(ana.name, ana.age)     # Ana 22
print(bo.name, bo.age)       # Bo 25, each object has its own data
```

## 3. Methods
A **method** is a function inside a class: something the object **can do**.
- It always takes `self` as its first parameter.
- It can **read** the object's data (`self.age`) and **change** it (`self.age += 1`).

```python
class Dog:
    def __init__(self, name):
        self.name = name
        self.energy = 5

    def bark(self):                     # an action that returns something
        return f"{self.name} says Woof!"

    def play(self):                     # an action that changes the data
        self.energy -= 1

rex = Dog("Rex")
print(rex.bark())
rex.play()
rex.play()
print(rex.energy)   # 3
```

## 4. Class attributes (shared data)
- `self.x` is data that belongs to **one object**.
- A variable written **directly inside the class** (not in `__init__`) is **shared by all objects**.

> 🏫 **Analogy:** each student has their own **notebook** (`self.name`), but the whole class shares one **whiteboard** (`school`).

```python
class Student:
    school = "IE University"     # shared by every student

    def __init__(self, name):
        self.name = name         # each student's own data

a = Student("Ana")
b = Student("Bo")
print(a.name, "-", a.school)
print(b.name, "-", b.school)
print(Student.school)            # you can also read it from the class itself
```

## 5. `__str__`: what `print()` shows
Without `__str__`, printing an object shows something ugly like `<__main__.Book object at 0x7f...>`. With `__str__`, **you** decide what the text looks like.

```python
class Book:
    def __init__(self, title, author):
        self.title = title
        self.author = author

    def __str__(self):
        return f"'{self.title}' by {self.author}"

b = Book("Dune", "Frank Herbert")
print(b)      # 'Dune' by Frank Herbert
```

## 6. Inheritance and `super()`
A **child class** gets everything from its **parent class** for free. It can then:
- **add** new things, and
- **override** (replace) a method by writing a new version with the same name.

`super().__init__(...)` runs the parent's `__init__`, so you don't have to repeat that setup code.

> 🧬 **Analogy:** a child inherits their parent's traits (eye color) but can also have their own (a different hobby). A `Cat` **is an** `Animal`, so it inherits everything an `Animal` has.

```python
class Animal:                          # parent
    def __init__(self, name):
        self.name = name

    def speak(self):
        return "..."

    def introduce(self):
        return f"I am {self.name} and I say {self.speak()}"

class Cat(Animal):                     # child: Cat IS AN Animal
    def speak(self):                   # override
        return "Miau"

class Dog(Animal):
    def __init__(self, name, breed):
        super().__init__(name)         # let the parent set self.name
        self.breed = breed             # add something new

    def speak(self):
        return "Woof"

print(Cat("Luna").introduce())
d = Dog("Rex", "Labrador")
print(d.introduce(), "| breed:", d.breed)
```

## ⚠️ Common beginner mistakes

| Mistake | Fix |
|---|---|
| Forgetting `self` as the first parameter of a method | Write `def bark(self):` |
| Writing `name = name` inside `__init__` | It must be `self.name = name`, otherwise the data is lost |
| Calling `rex.bark` without parentheses | `rex.bark()` |
| Forgetting `super().__init__(...)` in a child class | The parent's attributes never get created |

## 🎯 Challenge 1: `Pet` (practises lessons 2–3)
Create a `Pet` class:
- `Pet(name, species)` stores `name` and `species`, and sets `hunger = 5`.
- `feed()` lowers `hunger` by 1, but never below 0.
- `info()` returns `"Rex the dog (hunger: 5)"`.

```python
# ✅ ANSWER: Challenge 1
class Pet:
    def __init__(self, name, species):
        self.name = name
        self.species = species
        self.hunger = 5

    def feed(self):
        if self.hunger > 0:
            self.hunger -= 1

    def info(self):
        return f"{self.name} the {self.species} (hunger: {self.hunger})"
```

**💡 Why it works:** `__init__` saves the inputs with `self.` so they belong to this pet. `hunger = 5` is a starting value that isn't passed in. `feed()` uses an `if` guard so hunger can't go negative, and `info()` reads the object's own data with an f-string.

```python
def test_1():
    p = Pet("Rex", "dog")
    assert p.name == "Rex" and p.species == "dog", "save name and species with self."
    assert p.hunger == 5, "hunger should start at 5"
    assert p.info() == "Rex the dog (hunger: 5)", f"got {p.info()!r}"
    p.feed(); p.feed()
    assert p.hunger == 3, f"after 2 feeds hunger should be 3, got {p.hunger}"
    for _ in range(10): p.feed()
    assert p.hunger == 0, "hunger should never go below 0"
TESTS["1"] = test_1; check("Challenge 1", test_1)
```

## 🎯 Challenge 2: `Rectangle` (practises lesson 3)
- `Rectangle(width, height)`
- `area()` returns `width * height`.
- `perimeter()` returns `2 * (width + height)`.
- `is_square()` returns `True` when width equals height.

```python
# ✅ ANSWER: Challenge 2
class Rectangle:
    def __init__(self, width, height):
        self.width = width
        self.height = height

    def area(self):
        return self.width * self.height

    def perimeter(self):
        return 2 * (self.width + self.height)

    def is_square(self):
        return self.width == self.height
```

**💡 Why it works:** Every method reads `self.width` / `self.height`, the data saved in `__init__`. `is_square()` returns the comparison directly: `self.width == self.height` is already `True` or `False`, so `if ... return True else return False` isn't needed.

```python
def test_2():
    r = Rectangle(2, 3)
    assert r.area() == 6, f"area should be 6, got {r.area()}"
    assert r.perimeter() == 10, f"perimeter should be 10, got {r.perimeter()}"
    assert r.is_square() is False
    assert Rectangle(4, 4).is_square() is True
TESTS["2"] = test_2; check("Challenge 2", test_2)
```

## 🎯 Challenge 3: `Student` grades (practises lessons 3–4)
- Class attribute `school = "IE University"`.
- `Student(name)` starts with an **empty** list `self.grades = []`.
- `add_grade(grade)` adds the grade to the list.
- `average()` returns the average of the grades, or `0` if there are no grades yet.
- `passed()` returns `True` if the average is `>= 5`.

```python
# ✅ ANSWER: Challenge 3
class Student:
    school = "IE University"

    def __init__(self, name):
        self.name = name
        self.grades = []          # created inside __init__, so every student gets their own list

    def add_grade(self, grade):
        self.grades.append(grade)

    def average(self):
        if len(self.grades) == 0:
            return 0
        return sum(self.grades) / len(self.grades)

    def passed(self):
        return self.average() >= 5   # a method can call another method
```

**💡 Why it works:** `school` sits in the class body, so it's shared by every student. `self.grades = []` sits **inside `__init__`**, so each student gets their **own** list. The `if` in `average()` avoids dividing by zero. `passed()` reuses `average()`, which means methods can call each other through `self`.

```python
def test_3():
    s = Student("Vin")
    assert s.grades == [], "grades should start as an empty list"
    assert s.average() == 0, "average with no grades should be 0"
    s.add_grade(8); s.add_grade(6); s.add_grade(7)
    assert s.grades == [8, 6, 7]
    assert s.average() == 7, f"average should be 7, got {s.average()}"
    assert s.passed() is True
    t = Student("Bo"); t.add_grade(3)
    assert t.passed() is False
    assert t.grades == [3], "each student needs their OWN list"
    assert Student.school == "IE University" and s.school == "IE University"
TESTS["3"] = test_3; check("Challenge 3", test_3)
```

## 🎯 Challenge 4: `BankAccount` (practises lessons 3 and 5)
- `BankAccount(owner, balance=0)`
- `deposit(amount)` adds money to the balance.
- `withdraw(amount)`:
  - if there's enough money, subtract the amount and return `True`
  - otherwise, change nothing and return `False`
- `__str__` returns `"Vin: €100"`.

```python
# ✅ ANSWER: Challenge 4
class BankAccount:
    def __init__(self, owner, balance=0):
        self.owner = owner
        self.balance = balance

    def deposit(self, amount):
        self.balance += amount

    def withdraw(self, amount):
        if amount > self.balance:
            return False
        self.balance -= amount
        return True

    def __str__(self):
        return f"{self.owner}: €{self.balance}"
```

**💡 Why it works:** `balance=0` is a default parameter, so `BankAccount("Ana")` works. `withdraw()` checks **before** changing anything, so a failed withdrawal leaves the balance untouched. `__str__` controls what `print(acc)` and `str(acc)` show.

```python
def test_4():
    acc = BankAccount("Vin", 100)
    assert str(acc) == "Vin: €100", f"got {str(acc)!r}"
    acc.deposit(50)
    assert acc.balance == 150
    assert acc.withdraw(30) is True and acc.balance == 120
    assert acc.withdraw(500) is False, "not enough money -> False"
    assert acc.balance == 120, "a failed withdraw must not change the balance"
    assert BankAccount("Ana").balance == 0, "default balance is 0"
TESTS["4"] = test_4; check("Challenge 4", test_4)
```

## 🎯 Challenge 5: Animals (practises lesson 6, overriding)
- `Animal(name)` has:
  - `speak()`, which returns `"..."`
  - `introduce()`, which returns `"Rex says Woof"` by **using** `self.speak()`
- `Dog(Animal)` overrides `speak()` to return `"Woof"`.
- `Cat(Animal)` overrides `speak()` to return `"Miau"`.
- `Cow(Animal)` overrides `speak()` to return `"Muu"`.

Notice that `introduce()` is written **only once**, in the parent class, yet it works for every animal.

```python
# ✅ ANSWER: Challenge 5
class Animal:
    def __init__(self, name):
        self.name = name

    def speak(self):
        return "..."

    def introduce(self):
        return f"{self.name} says {self.speak()}"   # calls the CHILD's speak if it has one

class Dog(Animal):
    def speak(self):
        return "Woof"

class Cat(Animal):
    def speak(self):
        return "Miau"

class Cow(Animal):
    def speak(self):
        return "Muu"
```

**💡 Why it works:** `introduce()` is written once, in `Animal`. When it calls `self.speak()`, Python looks for `speak` on the actual object first (Dog, Cat or Cow) and only falls back to `Animal.speak` if the child doesn't have one. That's **overriding**, and it's why each child only needs one small method.

```python
def test_5():
    assert Animal("Thing").introduce() == "Thing says ..."
    assert Dog("Rex").introduce() == "Rex says Woof"
    assert Cat("Luna").introduce() == "Luna says Miau"
    assert Cow("Lola").introduce() == "Lola says Muu"
    assert isinstance(Dog("x"), Animal), "Dog must inherit from Animal"
    zoo = [Dog("A"), Cat("B"), Cow("C")]
    assert [a.speak() for a in zoo] == ["Woof", "Miau", "Muu"]
TESTS["5"] = test_5; check("Challenge 5", test_5)
```

## 🎯 Challenge 6: Vehicles (practises lesson 6, `super()`)
- `Vehicle(brand, wheels)` has `describe()`, which returns `"Seat with 4 wheels"`.
- `Car(brand, seats)` is a `Vehicle` with **4 wheels** and an extra `seats` attribute.
- `Motorcycle(brand)` is a `Vehicle` with **2 wheels**.

Use `super().__init__(...)` in both child classes.

```python
# ✅ ANSWER: Challenge 6
class Vehicle:
    def __init__(self, brand, wheels):
        self.brand = brand
        self.wheels = wheels

    def describe(self):
        return f"{self.brand} with {self.wheels} wheels"

class Car(Vehicle):
    def __init__(self, brand, seats):
        super().__init__(brand, 4)
        self.seats = seats

class Motorcycle(Vehicle):
    def __init__(self, brand):
        super().__init__(brand, 2)
```

**💡 Why it works:** `super().__init__(brand, 4)` lets the parent set `brand` and `wheels`, so that code isn't repeated. `Car` then adds its own `seats`. Neither child writes `describe()`; they inherit it from `Vehicle`.

```python
def test_6():
    c = Car("Seat", 5)
    assert c.describe() == "Seat with 4 wheels", f"got {c.describe()!r}"
    assert c.seats == 5
    m = Motorcycle("Yamaha")
    assert m.describe() == "Yamaha with 2 wheels"
    assert isinstance(c, Vehicle) and isinstance(m, Vehicle)
TESTS["6"] = test_6; check("Challenge 6", test_6)
```

## 🏆 Mini project: Shopping Cart (everything together)
**`Product(name, price)`**
- `__str__` returns `"Coffee - €2.5"`.

**`ShoppingCart()`** starts with an empty list of products:
- `add(product)` adds a `Product` to the cart.
- `total()` returns the sum of all the prices.
- `count()` returns the number of products.
- `most_expensive()` returns the most expensive `Product`, or `None` if the cart is empty.

*Hint:* a loop over `self.products` solves most of these.

```python
# ✅ ANSWER: Mini project
class Product:
    def __init__(self, name, price):
        self.name = name
        self.price = price

    def __str__(self):
        return f"{self.name} - €{self.price}"

class ShoppingCart:
    def __init__(self):
        self.products = []

    def add(self, product):
        self.products.append(product)

    def total(self):
        total = 0
        for p in self.products:
            total += p.price
        return total

    def count(self):
        return len(self.products)

    def most_expensive(self):
        if not self.products:
            return None
        best = self.products[0]
        for p in self.products:
            if p.price > best.price:
                best = p
        return best
```

**💡 Why it works:** `ShoppingCart` **holds** `Product` objects in a list, so it accesses each price as `p.price`. `total()` and `most_expensive()` are plain loops. The empty check in `most_expensive()` returns `None` before `self.products[0]` could crash. `self.products = []` in `__init__` gives each cart its own list.

```python
def test_project():
    coffee, bread, cheese = Product("Coffee", 2.5), Product("Bread", 1.2), Product("Cheese", 6.0)
    assert str(coffee) == "Coffee - €2.5", f"got {str(coffee)!r}"
    cart = ShoppingCart()
    assert cart.count() == 0 and cart.total() == 0
    assert cart.most_expensive() is None, "empty cart -> None"
    cart.add(coffee); cart.add(bread); cart.add(cheese)
    assert cart.count() == 3
    assert abs(cart.total() - 9.7) < 1e-9, f"total should be 9.7, got {cart.total()}"
    assert cart.most_expensive() is cheese
    assert ShoppingCart().count() == 0, "each cart needs its OWN list"
TESTS["project"] = test_project; check("Mini project", test_project)
```
